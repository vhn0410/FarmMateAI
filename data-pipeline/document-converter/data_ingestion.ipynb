```python
import os
import nest_asyncio
from pathlib import Path
from llama_parse import LlamaParse
from google.colab import userdata

# Lấy các key từ Colab Secrets (hoặc bạn có thể gán thẳng chuỗi string vào đây nếu làm nháp)

nest_asyncio.apply() # Bắt buộc trên Colab

# Thay bằng API Key của bạn
os.environ["LLAMA_CLOUD_API_KEY"] = userdata.get('LLAMA_CLOUD_API_KEY')

# ==========================================
# CẤU HÌNH ĐƯỜNG DẪN THƯ MỤC
# ==========================================
input_pdf_dir = '/content/drive/MyDrive/thesis/data/documents'
output_md_dir = '/content/drive/MyDrive/thesis/enterprise-data/new'
os.makedirs(output_md_dir, exist_ok=True)

# Khởi tạo LlamaParse tối ưu cho tài liệu học thuật
parser = LlamaParse(
    result_type="markdown",  # Ép xuất ra Markdown chuẩn
    premium_mode=True,       # Bật AI nhận diện Layout nâng cao (nhận diện chia cột)
    language="vi"            # Hỗ trợ tiếng Việt
)

pdf_files = list(Path(input_pdf_dir).glob('**/*.pdf'))
print(f"🔍 Tìm thấy {len(pdf_files)} file PDF cần chuyển đổi.\n")

for pdf_path in pdf_files:
    try:
        print(f"⏳ Đang xử lý bằng AI Vision: {pdf_path.name}...")

        # Gọi API Parse
        documents = parser.load_data(str(pdf_path))

        # Gộp toàn bộ các trang lại thành 1 chuỗi Markdown
        md_text = "\n\n".join([doc.text for doc in documents])

        output_filename = pdf_path.stem + ".md"
        output_path = os.path.join(output_md_dir, output_filename)

        with open(output_path, "w", encoding="utf-8") as f:
            f.write(md_text)

        print(f"✅ Đã lưu thành công: {output_filename}")

    except Exception as e:
        print(f"❌ Lỗi: {str(e)}")

print("\n🎉 Hoàn tất! File Markdown giờ đây đã chuẩn xác 100% cấu trúc.")
```
